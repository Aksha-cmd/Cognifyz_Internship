### Level 2, Task 1: Table Booking and Online Delivery Analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv('Dataset .csv')
# Normalize column names
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')

print("Dataset loaded successfully for Level 2!\n")
print('Columns:', list(df.columns))

# 1. Online delivery percentages
online_delivery_pct = df['has_online_delivery'].value_counts(normalize=True) * 100
print("--- Online Delivery Analysis ---")
print(f"Percentage offering Online Delivery: {online_delivery_pct.get('Yes', 0):.2f}%")
print(f"Percentage with NO Online Delivery:  {online_delivery_pct.get('No', 0):.2f}%\n")

# 2. Table booking percentages
table_booking_pct = df['has_table_booking'].value_counts(normalize=True) * 100
print("--- Table Booking Analysis ---")
print(f"Percentage offering Table Booking:   {table_booking_pct.get('Yes', 0):.2f}%")
print(f"Percentage with NO Table Booking:    {table_booking_pct.get('No', 0):.2f}%\n")

# 3. Average ratings comparison
df['aggregate_rating'] = pd.to_numeric(df['aggregate_rating'], errors='coerce')
avg_rating_with_booking = df[df['has_table_booking'] == 'Yes']['aggregate_rating'].mean()
avg_rating_no_booking = df[df['has_table_booking'] == 'No']['aggregate_rating'].mean()

print('--- Average Rating Comparison ---')
print(f"Average Rating WITH Table Booking:    {avg_rating_with_booking:.2f}")
print(f"Average Rating WITHOUT Table Booking: {avg_rating_no_booking:.2f}")
```

### Level 2, Task 2: Price Range Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Determine the most common price range
most_common_price = df['price_range'].mode()[0]
print("--- Most Common Price Range ---")
print(f"The most common price range designation is: {most_common_price}\n")

# 2. Calculate the average rating for each price range
avg_rating_by_price = df.groupby('price_range')['aggregate_rating'].mean().reset_index()
print("--- Average Rating by Price Range ---")
for index, row in avg_rating_by_price.iterrows():
    print(f"Price Range {int(row['price_range'])}: Average Rating = {row['aggregate_rating']:.2f}")
print("\n")

# 3. Analyze online delivery availability across different price ranges
# We create a cross-tabulation showing the percentage of delivery availability per price tier
delivery_by_price = pd.crosstab(df['price_range'], df['has_online_delivery'], normalize='index') * 100

print("--- Online Delivery Availability by Price Range (%) ---")
for price_tier in delivery_by_price.index:
    yes_pct = delivery_by_price.loc[price_tier, 'Yes']
    no_pct = delivery_by_price.loc[price_tier, 'No']
    print(f"Price Range {price_tier} -> Offers Delivery: {yes_pct:.2f}% | No Delivery: {no_pct:.2f}%")
    
```

### Level 2, Task 3: Feature Engineering and Categorical Encoding

```python
# 1. Feature Engineering: Create a new column for the length of the restaurant name
df['restaurant_name_len'] = df['restaurant_name'].apply(lambda x: len(str(x)))

print("--- Feature Engineering: Name Length Column ---")
print("New columns included in dataset:")
print(df[['restaurant_name', 'restaurant_name_len']].head(3))
print("\n")


# 2. Categorical Encoding: Convert 'Yes'/'No' text values into binary numbers (1 and 0)
# We will create clean encoded versions of our text tracking columns
df['table_booking_encoded'] = df['has_table_booking'].map({'Yes': 1, 'No': 0})
df['online_delivery_encoded'] = df['has_online_delivery'].map({'Yes': 1, 'No': 0})

print("--- Categorical Encoding: Binary Mapping ---")
print("Verification of successfully encoded variables:")
print(df[['has_table_booking', 'table_booking_encoded', 'has_online_delivery', 'online_delivery_encoded']].head(5))
```
